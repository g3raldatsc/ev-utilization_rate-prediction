# ****PERUMUSAN MASALAH****

---

INFORMASI SOAL :

Data Competition ISFEST 2026 bertujuan untuk mengetahui tingkat kemampuan para peserta dalam bidang data science dengan menyelesaikan studi kasus yang berjudul **Pemetaan Permintaan Infrastruktur Pengisian Daya Kendaraan Listrik (EV)** dengan tema yang bersangkutan dengan SDGs 7 : Affordable and Clean Energy.

<br>-

**HAL PENTING**

1. Membandingkan keberhasilan temuan menggunakan beberapa model prediksi untuk mendapatkan prediksi dari fenomena tertentu.
2. Melakukan analisis dan menetapkan penemuan faktor-faktor yang sangat berpengaruh terhadap tingkat penggunaan (utilization_rate) infrastruktur pengisian daya EV dan kesimpulan dari prediksi yang dilakukan.

---

PENCIPTA MASALAH :

1. Penggunaan EV meningkat, sehingga permintaan tempat pengisian daya meningkat.

MASALAH YANG DIHASILKAN :

1. Tidak merata dan kurangnya efisiensi pengalokasian tempat pengisian daya EV.

>Dari masalah tersebut dapat terjadi dari berbagai faktor seperti waktu, daerah, cuaca, harga, dan kualitas stasiun itu sendiri. Tanpa adanya sebuah analisis, masalah ini akan menjadi sulit untuk diselesaikan. Dengan memprediksi aspek yang mempengaruhi masalah ini dapat mengoptimalkan penempatan stasiun pengisian daya berenergi listrik dan perusahaan yang terlibat dapat terhindar dari kerugian.

DATASET YANG TERSEDIA :






>Dataset akan terbagi menjadi dua yaitu “train.csv” dan “test.csv”. “Train.csv” akan berisi semua kolom di atas disertai isi “utilization_rate”. Sementara “test.csv” akan berisi semua kolom di atas kecuali “utilization_rate”. Peserta akan diberikan “sample_submission.csv” sebagai panduan dalam mengumpulkan hasil prediksi pada leaderboard.

*Dalam pengolahan data, peserta dibebaskan menangani outliers pada fitur cuaca dan lalu lintas, namun sangat disarankan menggunakan fitur temporal untuk mengisi nilai kosong pada suhu dan indeks kemacetan. Peserta juga perlu mewaspadai anomali data; stasiun bernama sama dengan ID berbeda harus dianggap sebagai entitas terpisah, dan pastikan untuk mengecek kontinuitas waktu sebelum melakukan agregasi akibat adanya data gap yang signifikan pada tanggal 31 Desember 2025.*
