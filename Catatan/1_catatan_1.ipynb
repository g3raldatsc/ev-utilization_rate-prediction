#### ****Mapping the Demand for Electrical Vehicle (EV) Charging Infrastructure****

Seiring meningkatnya penggunaan kendaraan listrik, kebutuhan
stasiun pengisian daya juga ikut meningkat, namun ketersediaan
dan pemanfaatannya masih belum merata di beberapa lokasi. Dari
itu terjadi antrean panjang akibat tingginya permintaan, sementara
di lokasi lain justru kurang dimanfaatkan. Permintaan pengisian
daya dipengaruhi oleh berbagai faktor seperti waktu, kondisi lalu
lintas, cuaca, harga, dan acara lokal. Tanpa analisis yang tepat
terhadap faktor-faktor tersebut, sulit untuk memprediksi pola
permintaan dan memastikan penggunaan infrastruktur pengisian
daya yang efisien.

*KATA KUNCI : ANALISIS, MEMPREDIKSI POLA, EFISIEN*
