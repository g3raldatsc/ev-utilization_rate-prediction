## ****INFORMASI PENTING YANG WAJIB DIBACA****

---

Label 

>utilization_rate (Float) = Persentase hunian slot stasiun pengisian daya, yang dibatasi secara kontinu antara 0.0 sampai 1.0

**FITUR TEMPORAL DAN STRUKTURAL**

1. timestamp (DATETIME) = Kerangka waktu catatan log yang diambil setiap 30 menit. Catatan: Peserta harus memeriksa kontinuitas waktu sebelum agregasi karena adanya kesenjangan data yang signifikan pada tanggal 31 Desember 2025

**FITUR SPASIAL DAN INFRASTRUKTUR**

1. station_id (String ) = Pengidentifikasi unik yang diberikan kepada setiap stasiun pengisian daya. Peringatan Anomali: Stasiun dengan nama yang identik tetapi ID stasiun yang berbeda harus diperlakukan sebagai entitas fisik yang independen
2. station_name (String) = Nama spesifik yang sesuai dengan lokasi stasiun pengisian daya
3. network (String) = Penyedia jaringan yang mengendalikan operasi stasiun (misalnya, ChargePoint, Tesla)
4. city (String) = Kota spesifik tempat stasiun tersebut berada.
5. state (String) - Lokasi negara bagian atau provinsi yang terdaftar.
6. latitude/ longitude (Float) - Pemetaan koordinat geografis yang tepat dari lokasi penempatan stasiun.
7. location_type (String) - Jenis tempat yang menampung pengisi daya (misalnya, Mal, Parkir Umum, Bandara).
8. charger_type (String) - Standar konektor spesifik yang digunakan (misalnya, Level 2, DC Fast).
9. power_output_kw (Float) - Kapasitas transmisi daya listrik yang diukur dalam kilowatt (kW).
10. ports_total (Integer) - Jumlah absolut port pengisian daya yang terpasang di lokasi stasiun.
11. amenities_nearby (String) - Fasilitas komersial atau rekreasi yang dapat diakses di sekitar stasiun.
12. id

**FITUR HARGA DAN LINGKUNGAN**

1. pricing_type (String) - Mekanisme penagihan keuangan yang diterapkan (misalnya, Biaya Tetap, Per kWh).
2. temperature_f (Float) - Pembacaan suhu lingkungan luar ruangan lokal pada waktu pencatatan, dicatat dalam Fahrenheit. Catatan Nilai yang Hilang: Sangat disarankan untuk menggunakan properti temporal untuk mengisi data yang hilang dalam fitur ini.
3. precipitation_mm (Float) - Volume curah hujan atmosfer yang tercatat diukur dalam milimeter.
4. weather_condition (String) - Deskripsi kualitatif tentang kondisi cuaca (misalnya, Cerah, Hujan, Bersalju).
5. gas_price_per_gallon (Float) - Harga rata-rata bensin per galon di wilayah setempat, yang berfungsi sebagai indikator alternatif pasar acuan.
6. local_event (String) - Indikator log yang menunjukkan acara pertemuan khusus lingkungan jika sedang berlangsung (misalnya, Konser, Olahraga).

>Tugas ini memerlukan pembuatan model regresi (seperti Regresi Linier, Random Forest, atau Decision Tree Regressor) untuk memprediksi nilai kontinu yang tepat untuk variabel tersebut utilization_rate. Pengajuan diproses melalui sistem papan peringkat menggunakan metrik Root Mean Squared Error (RMSE).

---

- ****Anomali Entitas Stasiun****: Panitia secara eksplisit memperingatkan adanya stasiun dengan nama yang sama persis namun memiliki station_id yang berbeda. Stasiun-stasiun ini wajib diperlakukan sebagai entitas yang benar-benar terpisah, sehingga penggabungan data (merging atau groupby) tidak boleh sekadar bersandar pada nama lokasi

- ****Jebakan Kontinuitas Waktu (Time-Series Gap)****: Terdapat celah data yang signifikan pada tanggal 31 Desember 2025. Mengingat data dicatat setiap 30 menit, Anda harus sangat waspada melakukan pengecekan kontinuitas waktu sebelum mengeksekusi agregasi temporal (misalnya rata-rata harian atau mingguan) agar model tidak mempelajari pola pergeseran waktu yang salah

- ****Imputasi Cerdas Berbasis Waktu****: Terdapat instruksi khusus untuk penanganan nilai kosong (missing values) pada fitur suhu (temperature_f) dan indeks kemacetan lalu lintas (traffic_congestion_index). Panitia sangat menyarankan penggunaan fitur temporal (seperti jam, hari, atau bulan) untuk mengisi kekosongan ini, yang berarti metode imputasi rata-rata global sederhana (seperti SimpleImputer biasa) kemungkinan besar akan mengurangi nilai akurasi :3

- ****Sensitivitas Penanganan Outlier****: Peserta dibebaskan menangani outlier pada fitur cuaca dan lalu lintas. Tantangannya adalah membedakan mana pencilan yang merupakan noise murni (kesalahan sensor) dan mana yang merupakan kejadian ekstrem nyata (seperti badai salju atau kemacetan total) yang justru sangat mendikte lonjakan atau penurunan pada nilai target

- ****Pemodelan Regresi Spasial-Temporal****: Target utama adalah memprediksi utilization_rate, yakni persentase penggunaan stasiun dengan rentang nilai 0.0 hingga 1.0. Mengingat data memuat koordinat spasial, tipe lokasi, dan penanda jam sibuk, harus memetakan interaksi rumit antara lokasi fisik dan waktu. Evaluasi model pada sistem leaderboard akan dinilai murni menggunakan metrik Root Mean Squared Error (RMSE)

- ****Tuntutan Nilai Bisnis (Alur CRISP-DM)****: RMSE yang rendah tidak akan membawa tim lolos atau menang di babak final jika gagal diterjemahkan ke dalam bentuk actionable insight. Laporan akhir wajib memuat interpretasi model, penentuan faktor yang paling berpengaruh, serta saran strategis bagi pemerintah atau penyedia layanan yang relevan dengan target SDG
