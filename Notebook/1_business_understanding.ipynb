## ****BUSINESS UNDERSTANDING (PEMAHAMAN BISNIS)****

---

### ****LATAR BELAKANG****

Sejalan dengan Tujuan Pembangunan Berkelanjutan 7 (Energi Terjangkau dan Bersih) dan Tujuan 4 (Pendidikan Berkualitas), ISFEST 2026 Universitas Multimedia Nusantara menghadirkan tantangan berskala nasional yang berfokus pada masa depan infrastruktur transportasi hijau.

Seiring semakin cepatnya adopsi Kendaraan Listrik (Electric Vehicle/EV) di wilayah perkotaan, permintaan akan stasiun pengisian daya yang andal meningkat pesat. Namun, perencana kota dan operator swasta menghadapi hambatan efisiensi yang kritis: pemanfaatan stasiun sangat tidak merata di berbagai lokasi dan waktu. Di zona komersial berkepadatan tinggi, pengguna sering mengalami waktu tunggu yang lama akibat lonjakan permintaan. Sebaliknya, stasiun di daerah pinggiran kota atau permukiman kerap kurang dimanfaatkan, bahkan pada jam operasional puncak.

Memprediksi permintaan pengisian daya EV adalah masalah yang kompleks dan non-linear. Pola pemanfaatan dipengaruhi banyak faktor, di antaranya:

1. **Temporal**: waktu dalam sehari, hari dalam seminggu, dan dinamika musiman.
2. **Lingkungan**: kondisi cuaca, suhu, dan curah hujan.
3. **Spasial dan ekonomi**: lokasi dan tipe lokasi stasiun, acara lokal di sekitarnya, serta struktur harga (skema tarif dan harga bensin sebagai indikator pasar).

Tanpa peramalan berbasis data yang akurat, mengoptimalkan distribusi jaringan listrik dan menjamin kualitas layanan (QoS) bagi pengguna EV menjadi sulit dilakukan.

### ****PERUMUSAN TUJUAN BISNIS****

1. *Specific*: Memprediksi tingkat pemanfaatan (utilization_rate) setiap stasiun pengisian EV per 30 menit, agar operator dan pemerintah dapat menyeimbangkan pemanfaatan antarlokasi, menekan waktu tunggu di lokasi padat, dan menjaga QoS.
2. *Measurable*:
   - RMSE validasi (berbasis waktu) lebih rendah daripada baseline sederhana, yaitu rata-rata utilization_rate per stasiun per jam, serta seminimal mungkin pada leaderboard.
   - Laporan memuat interpretasi model, daftar faktor paling berpengaruh, dan minimal tiga rekomendasi strategis yang didukung temuan model.
3. *Achievable*: Tersedia 1.054.200 baris data train dengan 21 kolom (termasuk target) yang mencakup dimensi temporal, spasial, spesifikasi stasiun, harga, cuaca, dan acara lokal, serta data test sejumlah 263.550 baris dengan 20 kolom (tanpa target).
4. *Relevant*: Mendukung SDG 7 melalui efisiensi infrastruktur dan transisi energi bersih, serta SDG 4 melalui insight berbasis data yang dapat menjadi bahan edukasi dan dasar perencanaan kebijakan.
5. *Time-bound*: Seluruh alur CRISP-DM, laporan (PDF), dan materi presentasi (PPTX) selesai sebelum 25 September 2026 pukul 23.59 WIB.

### ****TUJUAN DATA MINING****

Membangun model regresi spasial-temporal yang memprediksi utilization_rate pada data test dengan RMSE serendah mungkin, kemudian menafsirkan model (feature importance) untuk menentukan faktor yang paling mendikte lonjakan dan penurunan pemanfaatan stasiun.

### ****PEMAHAMAN SINTESIS DATASET****

Dataset train berisi 1.054.200 baris dengan 21 kolom. Dataset test berisi 263.550 baris dengan 20 kolom (tanpa utilization_rate). Train mencakup 1 Juli 2025 00:00 sampai 24 November 2025 09:30 (7.028 timestamp), dan test mencakup 24 November 2025 10:00 sampai 31 Desember 2025 00:00 (1.757 timestamp), yang terkonfirmasi pada Data Understanding sebagai panel seimbang 150 stasiun tanpa celah waktu.

- *Pengenal baris*: id (tidak menjadi fitur prediksi).
- *Temporal*: timestamp (log per 30 menit).
- *Stasiun dan spasial*: station_id, station_name, network, city, state, latitude, longitude, location_type.
- *Spesifikasi dan harga*: charger_type, power_output_kw, ports_total, amenities_nearby, pricing_type.
- *Lingkungan dan ekonomi*: temperature_f, precipitation_mm, weather_condition, gas_price_per_gallon.
- *Konteks*: local_event.
- ***Target***: utilization_rate (kontinu, terpotong pada 0,02 sampai 0,98).

### ****KLASIFIKASI MASALAH MACHINE LEARNING****

Supervised Machine Learning: Regression Analysis. Target berupa variabel kontinu (utilization_rate, terpotong pada rentang 0,02 sampai 0,98) dengan struktur data panel spasial-temporal (150 stasiun, rekaman per 30 menit). Metrik evaluasi tunggal adalah Root Mean Squared Error (RMSE). Prediksi akhir dibatasi pada rentang 0,02 sampai 0,98, sesuai batas aktual pada data.

### ****PERNYATAAN MASALAH INTI (PROBLEM STATEMENT)****

Pemanfaatan stasiun pengisian EV sangat timpang: antrean panjang terjadi di lokasi dan jam tertentu, sementara stasiun lain sepi. Ketimpangan ini dipengaruhi interaksi kompleks antara waktu, lokasi dan tipe lokasi, spesifikasi stasiun, skema harga, cuaca, dan acara lokal yang belum terpetakan dengan akurat.

Pertanyaan yang dijawab:
1. Seberapa akurat utilization_rate dapat diprediksi dari faktor temporal, spasial, harga, cuaca, dan acara lokal?
2. Faktor apa yang paling berpengaruh terhadap utilization_rate?
3. Strategi apa yang dapat dilakukan pemerintah dan penyedia layanan untuk memeratakan pemanfaatan dan meningkatkan QoS?

### ****PEMANGKU KEPENTINGAN****

Perencana kota dan pemerintah (kebijakan dan infrastruktur), operator jaringan pengisian daya swasta (kapasitas, harga, operasional), penyedia listrik (beban jaringan), dan pengguna EV (waktu tunggu dan kualitas layanan).

### ****PENILAIAN SITUASI DAN BATASAN TEKNIS****

**1. Risiko dan anomali data**

| Risiko | Penanganan |
|---|---|
| Nama stasiun kembar dengan station_id berbeda (contoh dari gambaran data: Electrify America - Phoenix #11 dan Volta - San Diego #10) | Perlakukan sebagai entitas terpisah; gunakan station_id, bukan nama, untuk merge dan groupby |
| Celah data pada 31 Desember 2025 | Cek kontinuitas timestamp sebelum agregasi, lag, atau rolling |
| Nilai kosong pada temperature_f dan precipitation_mm | Imputasi berbasis fitur temporal (jam, hari, bulan) per kota, bukan rata-rata global |
| Outlier cuaca | Bedakan noise sensor dari kejadian ekstrem nyata (badai salju, panas ekstrem) yang justru mendikte target |
| Test berada setelah train, dan Desember tidak ada di train | Validasi berbasis waktu, bukan random split; waspadai pergeseran musim dan hari libur |
| Test tidak memiliki target | Fitur berbasis lag target tidak dapat dipakai langsung pada test |
| Kemungkinan inkonsistensi charger_type dan power_output_kw | Periksa pada Data Understanding |

**2. Aturan dan batasan lomba**
- Dilarang menambah dataset eksternal atau memanipulasi data mentah secara manual.
- Batas unggah submission leaderboard maksimal lima kali per hari.
- Evaluasi leaderboard murni menggunakan RMSE; penilaian akhir mencakup interpretasi dan saran strategis.

**3. Asumsi**
- station_id merepresentasikan satu stasiun fisik yang unik.
- Hubungan antarfitur dan target pada train berlaku pada test. Periode test terkonfirmasi berada pada rezim lingkungan yang identik dengan November di train sampai level kota, sehingga November menjadi acuan validasi utama.

**4. Perangkat dan teknologi**
- Python 3.13 pada virtual environment lokal dengan pandas, Scikit-Learn, LightGBM, XGBoost, dan CatBoost.

### ****RENCANA PROYEK (CRISP-DM)****

1. Business Understanding: tujuan, kriteria sukses, dan risiko.
2. Data Understanding: profil data, kontinuitas waktu, nilai kosong, outlier, dan stasiun kembar.
3. Data Preparation: imputasi berbasis waktu, rekayasa fitur temporal dan spasial, encoding.
4. Modeling: baseline, lalu model berbasis pohon dan boosting dengan validasi berbasis waktu.
5. Evaluation: RMSE, interpretasi model, faktor berpengaruh, dan saran strategis untuk SDG 7 dan SDG 4.
